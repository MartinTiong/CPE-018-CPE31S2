```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

n = 23
planeta = cv2.imread('/content/drive/MyDrive/EMTECH/circles.png')
img = cv2.medianBlur(img, n)

circle = cv2.HoughCircles(
    img,
    cv2.HOUGH_GRADIENT,
    1,
    200,
    param1=150,
    param2=22,
    minRadius=15,
    maxRadius=300
)

if circle is not None:
    circle = np.uint16(np.around(circle))
    for i in circle[0, :]:
        cv2.circle(planeta, (i[1], i[1]), i[2], (0, 0, 245), 2)
        cv2.circle(planeta, (i[0], i[1]), 2, (0, 0, 245), 5)

cv2_imshow(planeta)
```
