1. Create a Food class that defines the following:
* name of the food
* value of the food
* calories of the food

2. Create the following methods inside the Food class:
* A method that returns the value of the food
* A method that returns the cost of the food
* A method that calculates the density of the food (Value / Cost)
* A method that returns a string to display the name, value and calories of the food

```python
class Food(object):
    def __init__(self, n, v, w, p):
        # Make the variables private
        self.name = n
        self.value = v
        self.calories = w
        self.protein = p
    def getValue(self):
        return self.value
    def getCost(self):
        return self.calories
    def getProtein(self):
        return self.protein
    def density(self):
        return self.getValue()/self.getCost()
    def __str__(self):
        return self.name + ': <' + str(self.value)+ ', ' + str(self.calories) + '>'
```

3. Create a buildMenu method that builds the name, value and calories of the food


```python
def buildMenu(names, values, calories, protein):
    menu = []
    for i in range(len(values)):
        menu.append(Food(names[i], values[i],calories[i], protein[i]))
    return menu
```

4. Create a method greedy to return total value and cost of added food based on the desired maximum cost

```python
def greedy(items, maxCost, keyFunction):
    """Assumes items a list, maxCost >= 0,         keyFunction maps elements of items to numbers"""
    itemsCopy = sorted(items, key = keyFunction,
                       reverse = True)
    result = []
    totalValue, totalCost = 0.0, 0.0
    for i in range(len(itemsCopy)):
        if (totalCost+itemsCopy[i].getCost()) <= maxCost:
            result.append(itemsCopy[i])
            totalCost += itemsCopy[i].getCost()
            totalValue += itemsCopy[i].getValue()
    return (result, totalValue)
```

5. Create a testGreedy method to test the greedy method

```python
def testGreedy(items, constraint, keyFunction):
    taken, val = greedy(items, constraint, keyFunction)
    print('Total value of items taken =', val)
    for item in taken:
        print('   ', item)
```

```python
def testGreedys(foods, maxUnits):
    print('Use greedy by value to allocate', maxUnits,          'calories')
    testGreedy(foods, maxUnits, Food.getValue)
    print('\nUse greedy by cost to allocate', maxUnits,          'calories')
    testGreedy(foods, maxUnits, lambda x: 1/Food.getCost(x))
    print('\nUse greedy by density to allocate', maxUnits,          'calories')
    testGreedy(foods, maxUnits, Food.density)
    print('Use greedy by proteins to allocate', maxUnits,          'calories')
    testGreedy(foods, maxUnits, Food.getProtein)

```

9. Create method to use  Bruteforce algorithm instead of greedy algorithm

```python
def maxVal(toConsider, avail):
    """Assumes toConsider a list of items, avail a weight
       Returns a tuple of the total value of a solution to the
         0/1 knapsack problem and the items of that solution"""
    if toConsider == [] or avail == 0:
        result = (0, ())
    elif toConsider[0].getCost() > avail:
        #Explore right branch only
        result = maxVal(toConsider[1:], avail)
    else:
        nextItem = toConsider[0]
        #Explore left branch
        withVal, withToTake = maxVal(toConsider[1:],
                                     avail - nextItem.getCost())
        withVal += nextItem.getValue()
        #Explore right branch
        withoutVal, withoutToTake = maxVal(toConsider[1:], avail)
        #Choose better branch
        if withVal > withoutVal:
            result = (withVal, withToTake + (nextItem,))
        else:
            result = (withoutVal, withoutToTake)
    return result
```

```python
def testMaxVal(foods, maxUnits, printItems = True):
    print('Use search tree to allocate', maxUnits,
          'calories')
    val, taken = maxVal(foods, maxUnits)
    print('Total costs of foods taken =', val)
    if printItems:
        for item in taken:
            print('   ', item)
```

#### Supplementary Activity:

* Choose a real-world problem that solves knapsacks problem
* Use the greedy and brute force algorithm to solve knapsacks problem


variables: fare, walk distance, and mood
maximum budget for a week: 300
convenient walk distance before losing mood(5): 1000m
goal: maximize mood

There are 3 routes to take when commuting to the workplace:
Route 1: second shortest walk distance (500m), second highest fare (35) and longest wait, with barely any space left (mood: 15)
Route 2: Longest walk distance (1200m) with the lowest fare (20) and barely any wait and more available seats (mood: 22)
Route 3: book a ride, 0 walk distance but highest fare (200) and best mood (33)

```python
routes = [
    ("Route 1", 35, 500, 15),
    ("Route 2", 20, 1200, 22),
    ("Route 3", 200, 0, 33)
]
```

```python
MAX_BUDGET = 300
MAX_WALK = 1000
```

Bruteforce

```python
best_mood = 0
best_routes = []

n = len(routes)

for i in range(2 ** n):

    total_fare = 0
    total_walk = 0
    total_mood = 0
    selected_routes = []

    for j in range(n):

        if (i >> j) & 1:
            name, fare, walk, mood = routes[j]
            total_fare += fare
            total_walk += walk
            total_mood += mood
            selected_routes.append(name)

    if total_fare <= MAX_BUDGET and total_walk <= MAX_WALK:
        if total_mood > best_mood:
            best_mood = total_mood
            best_routes = selected_routes

print("Selected Routes:")
for r in best_routes:
    print( r)
```

Greedy

```python


# Sort routes by mood-to-fare ratio
routes_sorted = sorted(routes, key=lambda r: r[3] / r[1], reverse=True)

remaining_budget = MAX_BUDGET
remaining_walk = MAX_WALK
greedy_mood = 0
greedy_routes = []

for route in routes_sorted:
    name, fare, walk, mood = route

    if fare <= remaining_budget and walk <= remaining_walk:
        greedy_routes.append(name)
        remaining_budget -= fare
        remaining_walk -= walk
        greedy_mood += mood

# Display greedy result
print("Selected Routes:")
for r in greedy_routes:
    print("-", r)

print("Total Mood:", greedy_mood)
```
